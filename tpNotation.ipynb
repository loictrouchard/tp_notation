## calcul des moyennes d'UE

### pratie 1
notes et coeff saisie au clavier

```python
n = int(input("Combien de notes ? "))
total = 0
total_coeff = 0
for i in range(n):
   note = float(input("Entrer la note : "))
   coeff = float(input("Entrer le coefficient : "))
   total = total + note * coeff
   total_coeff = total_coeff + coeff
moyenne = total / total_coeff
print("La moyenne de l UE est : ", moyenne)
```

### pratie 2
notes et coeff dans une liste

```python
note=[]
coeff=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
total=0 #nombre total de note
totalc=0 #nombre total de coefficients
for i in range(0,21):
    note.append(int(input("quel est votre note:")))
    total+=note[i]*coeff[i]
    totalc+=coeff[i]
moyenne=total / totalc #total des notes diviser par le total des coefficients
print("la moyenne de l'UE est :",moyenne)
```
